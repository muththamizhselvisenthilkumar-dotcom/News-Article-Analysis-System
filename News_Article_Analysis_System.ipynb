```python
import nltk
import re

from nltk.sentiment import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()
```

```python
# Sample news article
article = """
The government has announced a new renewable energy project
to generate clean electricity. The project will create thousands
of jobs and help reduce pollution. Experts believe that this
initiative will support economic growth and protect the environment.
"""

# Sentiment Analysis
score = sia.polarity_scores(article)

if score['compound'] >= 0.05:
    sentiment = "Positive 😊"
elif score['compound'] <= -0.05:
    sentiment = "Negative 😞"
else:
    sentiment = "Neutral 😐"

# Word Count
words = re.findall(r'\b\w+\b', article)
word_count = len(words)

# Sentence Count
sentences = re.split(r'[.!?]+', article)
sentence_count = len([s for s in sentences if s.strip()])

# Character Count
character_count = len(article)

# Important keywords
keyword_list = [
    "government", "energy", "renewable", "electricity",
    "jobs", "pollution", "economy", "environment",
    "technology", "education", "health", "business"
]

found_keywords = []

for keyword in keyword_list:
    if keyword.lower() in article.lower():
        found_keywords.append(keyword)

# Display result
print("=" * 55)
print("             NEWS ARTICLE ANALYSIS SYSTEM")
print("=" * 55)

print("\nNEWS ARTICLE:")
print(article)

print("\n--- ARTICLE STATISTICS ---")
print("Word Count      :", word_count)
print("Sentence Count  :", sentence_count)
print("Character Count :", character_count)

print("\n--- SENTIMENT ANALYSIS ---")
print("Positive Score  :", round(score['pos'], 2))
print("Negative Score  :", round(score['neg'], 2))
print("Neutral Score   :", round(score['neu'], 2))
print("Compound Score  :", round(score['compound'], 2))
print("Overall Sentiment:", sentiment)

print("\n--- IMPORTANT KEYWORDS ---")
print(found_keywords if found_keywords else "No important keywords found")

print("\n--- CONCLUSION ---")

if sentiment.startswith("Positive"):
    print("The news article has a positive tone.")
elif sentiment.startswith("Negative"):
    print("The news article has a negative tone.")
else:
    print("The news article has a neutral tone.")

print("=" * 55)
```
